```python
# Ticker Class Analysis
#apple.info

# ------------- info section -----------------
apple.info
# we have the state and city of apple, the industry and the sector of apple. also amount of employees.

# ----------- history section ----------------
apple.history(start = date(2022, 3,1), end = date(2025,3,1), interval = "1mo")
# the data is every week from Tuesday to Saturday.
# except holidays - such as December and January (new year and christmas) which it isn't open.
#this is the main function that give 
# the columns are: open, high, Low, Close, Volume, Dividends, StockSplits
# Dividens are given each 3 months, in February, May, August, and November
# StockSplits is a rare event, in the last 3 years it didn't happen.

# ---------- actions section ------------------
apple.get_actions()
# this method gives us the days that Apple give dividents from the whole history, or made stock splits.

# ---------- capital gains section ------------
apple.get_capital_gains()
# none for apple

# ----------- news section ----------------------
apple.get_news()
# Json format for news about apple stock in yahoo!

apple.calendar
# this give us overview on events about apple's stock such as dividends and earnings date.

apple.get_sec_filings()
# this is apple's sec filings. each object in this list is a report that apple has published.
# we can use it the extract dates of apple's reports.

apple.get_recommendations()
# this is yahoo's reccomendation of the stock, based on last periods.

apple.get_upgrades_downgrades()
# this gives us df based of GradeDates,which known firms reccomended to buy / sell stock
# high value Data!!!

apple.get_earnings_history()
# for each q In the last year, what is the eps (estimate and actual)
# eps is a financial metric that shows how much profit a company generates 
# for each outstanding share of its common stock

apple.get_eps_trend()
# the trend of eps in periods - don't relevant this is the curr state.

apple.get_institutional_holders()
```
