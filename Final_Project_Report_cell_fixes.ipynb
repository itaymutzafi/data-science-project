To enrich each target ticker (AAPL, AMZN, GOOG, MSFT), this section adds two external blocks:

- **Peer-stock block**: for each target ticker, we merge the other three tech tickers' `Close`, `Volume`, and `Log_Return` (e.g., `Peer_AMZN_Close`, `Peer_MSFT_Volume`).

  - `Nasdaq_100` from `QQQ`
  - `Treasury_10Y` from `^TNX`

4.4.2

In this subsection, we incorporate a compact set of macro-financial variables to capture market-wide stress, technology-sector context, and interest-rate conditions.

Specifically, we merge four auxiliary ticker series (aligned by trading date) into each target stock frame: `VIX_Index` (from `^VIX`), `NVDA_Leader` (from `NVDA`), `Nasdaq_100` (from `QQQ`), and `Treasury_10Y` (from `^TNX`).

From `VIX_Index`, we further derive two regime-sensitive features: `VIX_MA20`, a 20-day moving average that represents the local volatility trend, and `VIX_Gap = VIX_Index - VIX_MA20`, which measures short-term deviation from that trend. Positive `VIX_Gap` values indicate elevated implied volatility relative to recent conditions, whereas negative values suggest comparatively calmer market sentiment.

Methodologically, the VIX itself is computed from S&P 500 options; therefore, S&P 500 information enters this pipeline indirectly through `VIX_Index`. Accordingly, `^GSPC`/`SPX` is not added as a separate direct feature in this section.
